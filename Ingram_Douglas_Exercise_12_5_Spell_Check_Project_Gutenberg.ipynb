 Exercise 12.5 – Spell Check a Project Gutenberg Book

This notebook downloads a Project Gutenberg book, creates a TextBlob, tokenizes the text into words, and checks the words for possible spelling errors and suggested corrections.

```python
!pip install textblob requests

import requests
import nltk
from textblob import TextBlob

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
url = "https://www.gutenberg.org/files/1661/1661-0.txt"

response = requests.get(url)
data = response.text

print(data[:1000])
```

```python
blob = TextBlob(data)

print("Number of characters:", len(data))
print("Number of words:", len(blob.words))
```

```python
words = blob.words

print(words[:50])
```

```python
unique_words = sorted(set(word.lower() for word in words if word.isalpha()))

print("Number of unique words:", len(unique_words))
print(unique_words[:50])
```

```python
from textblob import Word

possible_misspellings = []

for word in unique_words:
    suggestions = Word(word).spellcheck()
    suggestion, confidence = suggestions[0]

    if suggestion.lower() != word.lower():
        possible_misspellings.append(
            (word, suggestion, confidence)
        )

print("Possible misspellings and corrections:\n")

for word, suggestion, confidence in possible_misspellings[:25]:
    print(
        f"Original: {word:<15} "
        f"Suggestion: {suggestion:<15} "
        f"Confidence: {confidence:.3f}"
    )
```

Results: TextBlob identified several words that may be misspelled and provided suggested corrections with confidence values. Because the book contains names, older vocabulary, and proper nouns, some suggested corrections may not represent actual spelling errors.
