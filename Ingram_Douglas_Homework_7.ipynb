```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import time
```

# Exercise 16.1 - Fashion-MNIST vs. MNIST

This exercise compares the performance and training time of the convolutional neural network on the Fashion-MNIST and MNIST datasets.

```python
fashion_mnist = keras.datasets.fashion_mnist

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

X_train = X_train.astype("float32") / 255
X_test = X_test.astype("float32") / 255

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
from tensorflow.keras import models, layers

model = models.Sequential([
    layers.Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.summary()
```

## Fashion-MNIST Training

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start_time = time.time()

history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

training_time = time.time() - start_time

print("Training time:", training_time, "seconds")
```

```python
test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)

print("Fashion-MNIST test accuracy:", test_accuracy)
print("Fashion-MNIST training time:", training_time, "seconds")
```

# Exercise 16.4 - Convnet Layers

This exercise compares the original convolutional neural network with two modified versions: one with the first Dense layer removed and one with an added Dense layer containing 4096 neurons. The models are compared by test accuracy and training time.

```python
from tensorflow.keras import models, layers

model_no_dense = models.Sequential([
    layers.Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(10, activation='softmax')
])

model_no_dense.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
start_time = time.time()

history_no_dense = model_no_dense.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

no_dense_time = time.time() - start_time

print("Training time without first Dense layer:", no_dense_time, "seconds")
```

```python
no_dense_loss, no_dense_accuracy = model_no_dense.evaluate(
    X_test, y_test, verbose=0
)

print("Test accuracy without first Dense layer:", no_dense_accuracy)
print("Training time without first Dense layer:", no_dense_time, "seconds")
```

```python
model_4096 = models.Sequential([
    layers.Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(4096, activation='relu'),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model_4096.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
start_time = time.time()

history_4096 = model_4096.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

time_4096 = time.time() - start_time

print("Training time with 4096-neuron Dense layer:", time_4096, "seconds")
```

```python
loss_4096, accuracy_4096 = model_4096.evaluate(
    X_test, y_test, verbose=0
)

print("Test accuracy with 4096-neuron Dense layer:", accuracy_4096)
print("Training time with 4096-neuron Dense layer:", time_4096, "seconds")
```

Exercise 16.4 Results:
Removing the first Dense layer reduced the test accuracy from about 99.09% to 98.92%. The training time also changed from about 632.95 seconds to 627.84 seconds. Adding a Dense layer with 4096 neurons increased the test accuracy to about 99.17%, but it also increased the training time significantly to about 1774.24 seconds. The larger Dense layer produced only a small improvement in accuracy while requiring much more training time, so the original model provided a better balance between accuracy and speed.
